```python
import pandas as pd

csv_text = """State,Region,Doctors,Nurses,Population
Bihar,East,40000,30000,130000000
Maharashtra,West,150000,120000,125000000
Chhattisgarh,Central,12000,9000,30000000
Kerela,South,60000,90000,35000000
Odisha,East,25000,20000,46000000
Tamil Nadu,South,130000,140000,78000000
"""

with open("practice.csv", "w") as f:
    f.write(csv_text)
```

```python
df = pd.read_csv("practice.csv")
df.head()
```

```python
print(df.shape)
df.info()
print(df.isnull().sum())
```

```python
df["Doctors_per_1000"] = df["Doctors"] / df["Population"] * 1000
df["Nurses_per_1000"] = df["Nurses"] / df["Population"] * 1000
df.round(2)
```

```python
df.groupby("Region")[["Doctors_per_1000", 'Nurses_per_1000']].mean().round(2)
```

```python
df.sort_values("Nurses_per_1000")[df["Nurses_per_1000"] < 2]
```
