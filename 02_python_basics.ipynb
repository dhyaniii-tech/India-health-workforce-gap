```python
import pandas as pd
```

```python
data = {
    "State":  ["Bihar", "Maharashtra", "Chhattisgarh", "Kerela"],
    "Doctors": [40000, 150000, 12000, 60000],
    "Population": [130000000, 125000000, 30000000, 35000000]
}
df = pd.DataFrame(data)
df
```

```python
df["Doctors_per_1000"] = df["Doctors"] / df["Population"] * 1000
df
```

```python
df.sort_values("Doctors_per_1000")
```

```python
print(df.shape)
print(df.columns)
df.describe()
```
